## Exercise 1: Grade classifier

Write a function `classify_grade(grade)` that takes a numeric grade (from 0 to 10) and returns a string with the corresponding classification, following these criteria:

| Range | Classification |
|---|---|
| grade >= 9 | "Excellent" |
| 7 <= grade < 9 | "Good" |
| 5 <= grade < 7 | "Pass" |
| grade < 5 | "Fail" |

Then, using a **`for` loop**, apply the function to the following list of grades and store the results in a new list:

```python
grades = [3.5, 9.2, 6.8, 10, 4.9, 7.0, 5.0]
```

Finally, compute and print:
1. The **average** of the grades (you can use `numpy`).
2. **How many** students passed (grade >= 5).

**Hint:** remember how `if`/`elif`/`else` can be combined with loops (seen in the notebook with the Celsius-to-Fahrenheit conversion and with the sum of positive numbers).

````python
def classify_grade(grade):
    # YOUR CODE HERE
    if (grade >= 9):
        return "Excellent"
    elif (grade >= 7):
        return "Good"
    elif (grade >= 5):
        return "Pass"
    else:
        return "Fail"

grades = [3.5, 9.2, 6.8, 10, 4.9, 7.0, 5.0]

# YOUR CODE HERE: apply classify_grade to each element of 'grades'
results = [classify_grade(grade) for grade in grades]

# YOUR CODE HERE: compute the average with numpy
average = sum(grades) / len(grades)

# YOUR CODE HERE: count how many students passed
passed = len(grades) - results.count("Fail")

print(results)
print("Average:", average)
print("Passed:", passed)
````

## Exercise 2: Text processing and word-frequency dictionary

Given the following text:

```python
text = "python is a very powerful language and python is easy to learn"
```

1. Use the `.split()` method (seen in the notebook) to split the text into a list of words.
2. Loop through the list with a `for` loop and build a **dictionary** `frequencies` where each key is a word and each value is the number of times it appears in the text (hint: use the `in` operator to check whether the key already exists in the dictionary, as shown in the dictionaries section).
3. Write a function `most_frequent_word(frequencies)` that takes the dictionary and returns the word (key) with the highest frequency.
4. Print the full dictionary and the most frequent word.

**Bonus (optional):** use string *slicing* to print only the first 3 characters of the most frequent word.

````python
text = "python is a very powerful language and python is easy to learn"

# YOUR CODE HERE: split the text into words
words = text.split(" ")

# YOUR CODE HERE: build the frequency dictionary
frequencies = {}
for word in words:
    if word in frequencies.keys():
        frequencies[word] += 1
    else:
        frequencies[word] = 1

def most_frequent_word(frequencies: dict):
    # YOUR CODE HERE
    word = ""
    freq = 0
    for key, value in frequencies.items():
        if value > freq:
            freq = value
            word = key
    return word

print(frequencies)
print("Most frequent word:", most_frequent_word(frequencies))
````

## Exercise 3: Prime numbers + analysis with pandas

**Part A — Algorithm**

Write a function `is_prime(n)` that returns `True` if `n` is a prime number and `False` otherwise (take inspiration from how `factorial(n)` and `gcd(a,b)` were built in the notebook, using loops and conditionals).

Use this function together with a loop to generate the list of all prime numbers between 2 and 100, and store it in the variable `primes`.

**Part B — Data with pandas**

Create a pandas `DataFrame` with two columns, `number` and `is_prime`, for all numbers from 2 to 100 (`is_prime` column with `True`/`False` values using your function). You can build the DataFrame like this:

```python
import pandas as pd
df = pd.DataFrame({'number': list(range(2, 101))})
```

From that DataFrame:
1. Add the `is_prime` column by applying your function to the `number` column.
2. Filter the DataFrame to show only the prime numbers (`df[df['is_prime']]`).
3. Print how many prime numbers there are between 2 and 100.

**Hint:** to apply a function to each element of a pandas column you can use `.apply()`, for example: `df['number'].apply(is_prime)`.

````python
from math import sqrt

def is_prime(n):
    # YOUR CODE HERE
    for i in range(2, int(sqrt(n)) + 1):
        if (n % i == 0): return False
    return True

# YOUR CODE HERE: list of primes between 2 and 100
primes = [i for i in range(2, 101) if is_prime(i)]

print(primes)

import pandas as pd
df = pd.DataFrame({'number': list(range(2, 101))})

# YOUR CODE HERE: add the 'is_prime' column
df.insert(0, "is_prime", df["number"].apply(is_prime))
# YOUR CODE HERE: filter the DataFrame to keep only the primes
df_primes = df[df.is_prime]["number"]

print(df_primes)
print("Number of primes between 2 and 100:", df_primes.size)  # replace None with the correct calculation
````
